```python
import pandas as pd
```

```python
base_ano_meta = pd.read_excel('dados_exportados_jira_por_ano.xlsx', sheet_name='Sheet1')
base_metas_apuracao = pd.read_excel('dados_exportados_jira.xlsx', sheet_name='Sheet1')
```

```python
relacao_ano_meta = base_ano_meta[['Responsável', 'Ano da Meta']]
```

```python
relacao_ano_meta[relacao_ano_meta['Responsável'] == 'ASPLAGMETA-2772']
```

```python
# map 'Ano da Meta' from relacao_ano_meta to base_metas_apuracao where Chave == Responsável
mapping = relacao_ano_meta.set_index('Responsável')['Ano da Meta']
base_metas_apuracao['Ano da Meta'] = base_metas_apuracao['Chave'].map(mapping).fillna(base_metas_apuracao['Ano da Meta'])
```

```python
base_metas_apuracao[['Ano da Meta', 'META_ID', 'Responsável', 'Chave', 'Resumo']].head(14)
```

```python
# preencher valores ausentes de "Ano da Meta" em base_metas_apuracao usando o valor existente para o mesmo META_ID
na_before = base_metas_apuracao['Ano da Meta'].isna().sum()

base_metas_apuracao['Ano da Meta'] = base_metas_apuracao['Ano da Meta'].fillna(
    base_metas_apuracao.groupby('META_ID')['Ano da Meta'].transform(
        lambda x: x.dropna().iloc[0] if x.dropna().any() else float('nan')
    )
)

na_after = base_metas_apuracao['Ano da Meta'].isna().sum()
print(f"Ano da Meta: NaN antes={na_before}, depois={na_after}")
```
